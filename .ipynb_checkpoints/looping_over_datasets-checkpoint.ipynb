```python
with open('./data\\gapminder_gdp_oceania.csv') as file:
    print(file.readline())
```

```python
with open('./data\\gapminder_gdp_oceania.csv') as file:
    header = file.readline()
    table = []
    for line in file:
        table.append(line)
```

```python
collect = []
for row in table:
    c = row.split(sep=",")
    collect.append(float(c[3]))
```

combine above steps to calculate average

```python
collect = []
for row in table:
    c = row.split(sep=",")
    collect.append(float(c[3]))
sum(collect)/len(collect)
```
