## Turn it into a Function

```python
def average_for_a_column(col,data):
    collect = []
    for row in data:
        collect.append(float(row[col]))
    return sum(collect)/len(collect)
```

```python
# Function to read CSV file
def read_csv_file(filepath):
    with open(filepath, 'r') as f:
        lines = f.readlines()
    
    # Clean and split each line
    lines = [line.strip().split(',') for line in lines]
    headers = lines[0]   # first row (column names)
    data = lines[1:]     # rest of the rows
    return headers, data

```

```python
headers, data=read_csv_file("data/gapminder_gdp_americas.csv")
```

```python
print(average_for_a_column(data=data, col = 4))
```
