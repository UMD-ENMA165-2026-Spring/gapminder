## All Data

```python
data = pd.read_csv('data/gapminder_all.csv') 
data
```

```python
data.columns
```

```python
pd.wide_to_long(data, stubnames=['gdpPercap', 'pop', 'lifeExp'], i='country', j='year', sep='_')
```

```python
new_data = pd.wide_to_long(data, stubnames=['gdpPercap', 'pop', 'lifeExp'], i=['continent', 'country'], j='year', sep='_')
```

```python
new_data
```

```python
new_data.index
```

```python
new_data.loc['Africa']
```

```python
new_data.loc['Algeria']
```

```python
new_data.loc[('Africa', 'Algeria')]
```

```python
new_data = new_data.reset_index()
```

```python
continent = new_data['continent']=='Africa'
year = new_data['year']==1957
new_data.loc[continent&year]
```

```python
for i, group in new_data.groupby(by='continent'): 
    ax=None 
    for j, country in group.groupby(by='country'): 
        ax = country.plot('year', 'gdpPercap', title=i, ax=ax, label=j)
```
