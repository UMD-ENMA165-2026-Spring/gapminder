```python
oceania = pd.read_csv("./data/gapminder_gdp_oceania.csv", index_col="country")
oceania # conventionally df(data frame)
```

```python
oceania.plot()
```

```python
oceania.T.plot(use_index=False)
```

```python
oceania.T.plot(x="Australia", use_index=True)
```

```python
oceania.columns
```

```python
oceania.columns.str.split('_')
```

```python
def year_of_index(value):
    return int(value[1])
# year_of_index = lambda x : int(x[1])
yearsIndex = oceania.columns.str.split('_').map(year_of_index)
```

```python
oceania.columns = yearsIndex
oceania
```

```python
oceania.T.plot(xlabel="Year", ylabel="GDP in USD per capita", title="Oceania")
```

```python
oceania[1967]
```

```python
country = "Australia"
oceania.T[country].plot(xlabel="Year", ylabel="GDP in USD per capita", title=country)
```

```python
africa = pd.read_csv("./data/gapminder_gdp_africa.csv", index_col="country")
africa
```

```python
africa.columns.str.split('_').str[1].astype(int)
```

```python
aInAfrica = africa.index < "B"
def headers_to_years(data):
    return data.columns.str.split('_').map(year_of_index)
africa.columns = headers_to_years(africa)

africa[aInAfrica].T.plot()
```
